```python
# Import required libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
```

```python
# 1. Create a NumPy array with values from 10 to 50 (with step 5)
arr = np.arange(10, 55, 5)
print("Array:", arr)
```

```python
# 2. Slice the array to get elements from index 2 to 5
sliced = arr[2:6]   # index 2,3,4,5
print("Sliced (index 2 to 5):", sliced)
```

```python
# 3. Calculate the mean and standard deviation of the array
print("Mean:", np.mean(arr))
print("Standard Deviation:", np.std(arr))
```

```python
# 1. Create a Series with names of 5 fruits and their prices
fruits = pd.Series(
    {'apple': 1.2, 'banana': 0.5, 'cherry': 3.0, 'date': 2.5, 'elderberry': 4.0}
)
print(fruits)
```

```python
# 2. Access the price of the 3rd fruit by position
third_price = fruits.iloc[2]
print("Price of 3rd fruit (cherry):", third_price)
```

```python
# 3. Increase all prices by 10% (multiply by 1.1)
prices_increased = fruits * 1.1
print("Prices after 10% increase:\n", prices_increased)
```

```python
# 4. Calculate average price and find the most expensive fruit
avg_price = fruits.mean()
most_expensive = fruits.idxmax()
most_expensive_price = fruits.max()

print("Average price:", round(avg_price, 2))
print("Most expensive fruit:", most_expensive, "->", most_expensive_price)
```

```python
# 1. Create a DataFrame with 5 students and their exam scores in 3 subjects
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
    'Math': [85, 72, 90, 88, 65],
    'English': [78, 88, 92, 70, 80],
    'Science': [90, 75, 85, 82, 78]
}
df = pd.DataFrame(data)
df
```

```python
# 1. Create a DataFrame with sales data
data = {
    'Product': ['A', 'B', 'C', 'A', 'B', 'C', 'A', 'B', 'C', 'A', 'B', 'C'],
    'Region': ['North', 'South', 'East', 'West', 'North', 'South',
               'East', 'West', 'North', 'South', 'East', 'West'],
    'Sales': [100, 150, 200, 120, 180, 90, 160, 140, 210, 130, 170, 110]
}
df = pd.DataFrame(data)
df
```

```python
# 2. Group by Product and calculate the total sales for each product
df.groupby('Product')['Sales'].sum()
```

```python
# 3. Group by Region and find the average sales per region
df.groupby('Region')['Sales'].mean()
```

```python
# 4. Calculate both mean and max sales by Product
df.groupby('Product')['Sales'].agg(['mean', 'max'])
```

```python
# 1. Create a DataFrame with quarterly revenue data for 4 quarters
data = {
    'Quarter': ['Q1', 'Q2', 'Q3', 'Q4'],
    'Revenue': [12000, 15000, 18000, 22000]
}
df = pd.DataFrame(data)
df
```

```python
# 2. Line plot showing revenue over quarters
df.plot(x='Quarter', y='Revenue', kind='line', marker='o', figsize=(8, 5))
plt.title('Quarterly Revenue')
plt.xlabel('Quarter')
plt.ylabel('Revenue ($)')
plt.show()
```

```python
# 3. Bar chart comparing revenue across quarters
df.plot(x='Quarter', y='Revenue', kind='bar', figsize=(8, 5), color='skyblue')
plt.title('Revenue by Quarter')
plt.xlabel('Quarter')
plt.ylabel('Revenue ($)')
plt.xticks(rotation=0)
plt.show()
```

```python
# 4. DataFrame with student test scores and plot a histogram
scores_df = pd.DataFrame({
    'Score': [45, 50, 65, 70, 75, 80, 85, 90, 95, 88, 92, 78, 60, 82, 70]
})

scores_df['Score'].plot(kind='hist', bins=6, figsize=(8, 5), edgecolor='black')
plt.title('Distribution of Test Scores')
plt.xlabel('Score')
plt.ylabel('Frequency')
plt.show()
```

```python
# Load the Iris dataset from a URL
url = 'https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv'
iris_df = pd.read_csv(url)

iris_df.head()
```

```python
# Basic info
iris_df.shape
iris_df.info()
iris_df.describe()
```

```python
# 1. Filter to show only 'virginica' species - first 5 rows
virginica = iris_df[iris_df['species'] == 'virginica']
virginica.head()
```

```python
# 2. Find minimum and maximum petal width in the entire dataset
print("Min petal width:", iris_df['petal_width'].min())
print("Max petal width:", iris_df['petal_width'].max())
```

```python
# 3. Which species has the highest average sepal length?
avg_sepal_length = iris_df.groupby('species')['sepal_length'].mean()
print(avg_sepal_length)
print("\nSpecies with highest average sepal length:", avg_sepal_length.idxmax())
```

```python
# 4. Scatter plot: petal_length vs petal_width
iris_df.plot(x='petal_length', y='petal_width', kind='scatter', figsize=(8, 5))
plt.title('Petal Length vs Petal Width')
plt.xlabel('Petal Length (cm)')
plt.ylabel('Petal Width (cm)')
plt.show()
```

```python
# 5. Bar chart: average petal_length by species
iris_df.groupby('species')['petal_length'].mean().plot(
    kind='bar', figsize=(8, 5), color=['#4CAF50', '#2196F3', '#FF9800']
)
plt.title('Average Petal Length by Species')
plt.xlabel('Species')
plt.ylabel('Average Petal Length (cm)')
plt.xticks(rotation=0)
plt.show()
```
