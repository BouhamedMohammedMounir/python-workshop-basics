```python

import numpy as np

arr = np.arange(10, 51, 5)
print("Array:", arr)

print("Sliced array:", arr[2:5])

print("Mean:", np.mean(arr))
print("Standard deviation:", np.std(arr))

matrix = np.array([[1, 2, 3],
                   [4, 5, 6]])

print("Matrix:")
print(matrix)

print("Element at row 1, column 2:", matrix[1, 2])
```

Workbook practice 2

```python

import pandas as pd

fruits = pd.Series(
    [20, 35, 15, 40, 25],
    index=["Apple", "Banana", "Orange", "Mango", "Grapes"]
)

print("Original prices:")
print(fruits)

print("\nPrice of the 3rd fruit:", fruits.iloc[2])

fruits = fruits * 1.1

print("\nPrices after 10% increase:")
print(fruits)

print("\nAverage price:", fruits.mean())

print("Most expensive fruit:", fruits.idxmax())
print("Highest price:", fruits.max())
```

Workbook practice 3

```python

import pandas as pd

data = {
    "Student": ["Amira", "Sara", "Ahmed", "Yacine", "Fatima"],
    "Math": [90, 75, 88, 95, 82],
    "English": [92, 80, 86, 78, 90],
    "Science": [85, 90, 79, 88, 84]
}

df = pd.DataFrame(data)

print("1. All students:")
print(df)

print("\n2. First 3 rows:")
print(df.head(3))

print("\n3. Math scores:")
print(df["Math"])

print("\n4. Students with Math > 80:")
print(df[df["Math"] > 80])

print("\n5. Students with Math > 85 AND English > 85:")
print(df[(df["Math"] > 85) & (df["English"] > 85)])
```

 Workbook practice 4

```python

import pandas as pd

data = {
    "Product": ["A", "B", "A", "C", "B", "A", "C", "B"],
    "Region": ["North", "South", "East", "West",
               "North", "South", "East", "West"],
    "Sales": [100, 200, 150, 300, 250, 120, 350, 180]
}

df = pd.DataFrame(data)

print("1. Sales Data:")
print(df)

print("\n2. Total sales by Product:")
print(df.groupby("Product")["Sales"].sum())

print("\n3. Average sales by Region:")
print(df.groupby("Region")["Sales"].mean())

print("\n4. Mean and maximum sales by Product:")
print(df.groupby("Product")["Sales"].agg(["mean", "max"]))
```

```python

import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Quarter": ["Q1", "Q2", "Q3", "Q4"],
    "Revenue": [1000, 1500, 1200, 2000]
}

df = pd.DataFrame(data)

print("Quarterly Revenue:")
print(df)

plt.plot(df["Quarter"], df["Revenue"], marker="o")
plt.title("Revenue Over Quarters")
plt.xlabel("Quarter")
plt.ylabel("Revenue")
plt.show()

plt.bar(df["Quarter"], df["Revenue"])
plt.title("Revenue by Quarter")
plt.xlabel("Quarter")
plt.ylabel("Revenue")
plt.show()

scores_data = {
    "Student": ["Amira", "Sara", "Ahmed", "Fatima", "Yacine",
                "Meriem", "Omar", "Nour", "Ali", "Salim"],
    "Score": [55, 60, 65, 70, 70, 75, 80, 85, 90, 95]
}

scores_df = pd.DataFrame(scores_data)

print("Student Scores:")
print(scores_df)

plt.hist(scores_df["Score"], bins=5, edgecolor="black")
plt.title("Distribution of Student Scores")
plt.xlabel("Score")
plt.ylabel("Frequency")
plt.show()
```

```python

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("iris")

virginica = df[df["species"] == "virginica"]
print("1. First 5 virginica rows:")
print(virginica.head(5))

print("\n2. Minimum petal width:", df["petal_width"].min())
print("Maximum petal width:", df["petal_width"].max())

average_length = df.groupby("species")["sepal_length"].mean()
print("\n3. Average sepal length by species:")
print(average_length)
print("Highest average:", average_length.idxmax())

plt.scatter(df["petal_length"], df["petal_width"])
plt.title("Petal Length vs Petal Width")
plt.xlabel("Petal Length")
plt.ylabel("Petal Width")
plt.show()

average_petal = df.groupby("species")["petal_length"].mean()
average_petal.plot(kind="bar")
plt.title("Average Petal Length by Species")
plt.xlabel("Species")
plt.ylabel("Average Petal Length")
plt.show()
```
