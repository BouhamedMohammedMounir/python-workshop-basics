```python
rgb = ["R", "G", "B"]
rgb_copy = rgb          # NOT a real copy, same reference!
rgb_copy.append("A")
print(rgb_copy)         # ['R', 'G', 'B', 'A']
print(rgb)              # ['R', 'G', 'B', 'A']  -> also changed!
print(id(rgb) == id(rgb_copy))   # True -> same object
```

```python
rgb = ["R", "G", "B"]
correct_rgb_copy = rgb[:]        # real copy via slicing
correct_rgb_copy.append("A")
print(correct_rgb_copy)          # ['R', 'G', 'B', 'A']
print(rgb)                       # ['R', 'G', 'B']  -> unchanged
print(id(rgb) == id(correct_rgb_copy))   # False -> different objects
```

```python
# Sample list
numbers = [-3, 5, -1, 7, 0, -8, 10]

# Count positive numbers
count = 0
for n in numbers:
    if n > 0:
        count += 1

print("Count of positive numbers:", count)
# Output: Count of positive numbers: 3
```

```python
def filter_even_numbers(numbers):
    """
    Returns a list of even numbers from a list.
    """
    result = []
    for n in numbers:
        if n % 2 == 0:
            result.append(n)
    return result

test = [1, 2, 3, 4, 5, 6, 7, 8]
print(filter_even_numbers(test))
# Output: [2, 4, 6, 8]
```

```python
# Define lists
students = ["Alice", "Bob", "Charlie", "David", "Eve"]
grades = [85, 90, 78, 92, 90]

# Remove duplicates from grades list
unique_grades = set(grades)

# Find the highest grade
highest_grade = max(grades)

# Create the dictionary
student_grades = dict(zip(students, grades))

# Find list of student(s) who got the highest grade
top_students = [s for s, g in student_grades.items() if g == highest_grade]

# Print results
print("Student Grades Dictionary:", student_grades)
print("Unique Grades Set:", unique_grades)
print("Highest Grade:", highest_grade)
print("Top Student(s):", top_students)
```

```python
Student Grades Dictionary: {'Alice': 85, 'Bob': 90, 'Charlie': 78, 'David': 92, 'Eve': 90}
Unique Grades Set: {85, 90, 78, 92}
Highest Grade: 92
Top Student(s): ['David']
```

```python
%%writefile mymodule.py
def create_dict(keys, values):
    """
    Create a dictionary from two lists (keys and values).
    Returns an error message if lengths differ.
    """
    if len(keys) != len(values):
        return "Error: Lists must have the same length."
    return dict(zip(keys, values))
```
